```python
#Sayısal türev (numerical derivative) ile basit bir gradient descent döngüsü kur: parametreyi küçük adımlarla güncelleyerek loss'u düşür. (2. ve 3. video)

# karpatyın yaptığı videoda  h kdar sağa kaydırarak lossun ne kadar değiştiğine bakıyordu -> f'(w) = (f(w+h) - f(w)) / h
# araştırdığım merkezi fark ise hem sağa hem sola bakarak ilerliyor ikisi arasındaki farkı alıp 2h a bolüyor. -> f'(w) = (f(w+h) - f(w-h)) / (2h)
#ikisini de uygulayarak bakalım
```

```python
EULER = 2.718281828459045

def sigmoid(z):
    return 1 / (1 + (EULER ** -z))

def mse(y_true, y_pred):
    return (y_true - y_pred) ** 2

#  out = sigmoid(w*x + b), x ve b sabit, sadece w'yi optimize ediyoruz
x = 1.5
b = 0.2
y_true = 0.9

def forward(w):
    return sigmoid(w * x + b)

def loss_fn(w):
    return mse(y_true, forward(w))


def numerical_gradient(f, w, h=0.0001):
    return (f(w + h) - f(w - h)) / (2 * h) 

def numerical_gradient_karpathy(f, w, h=0.0001):
    return (f(w + h) - f(w)) / h

if __name__ == "__main__":
    learning_rate = 0.5
    steps = 40

    print("=== YONTEM 1: Merkezi fark (central difference) ===")
    w = -2.0
    print(f"{'adim':>4} | {'w':>8} | {'out':>8} | {'loss':>10} | {'grad':>10}")
    print("-" * 62)
    for step in range(steps):
        z = w * x + b
        out = sigmoid(z)
        loss = loss_fn(w)
        grad = numerical_gradient(loss_fn, w)

        print(f"{step:>4} | {w:>8.4f} | {out:>8.4f} | {loss:>10.6f} | {grad:>10.4f}")

        # grad pozitifse -> w'yi artirmak loss'u ARTIRIR -> w'yi AZALT
        # grad negatifse -> w'yi artirmak loss'u AZALTIR -> w'yi ARTIR
        
        w = w - learning_rate * grad

    print("-" * 62)
    print(f"Final w = {w:.4f}, final loss = {loss_fn(w):.6f}")
    print(f"Bu w ile noronun ciktisi: {forward(w):.4f}  (hedef: {y_true})")

    print()
```

```python
print("=== YONTEM 2: Karpathy usulu (forward difference) ===")
w = -2.0
print(f"{'adim':>4} | {'w':>8} | {'out':>8} | {'loss':>10} | {'grad':>10}")
print("-" * 62)
for step in range(steps):
    z = w * x + b
    out = sigmoid(z)
    loss = loss_fn(w)
    grad = numerical_gradient_karpathy(loss_fn, w)

    print(f"{step:>4} | {w:>8.4f} | {out:>8.4f} | {loss:>10.6f} | {grad:>10.4f}")

    w = w - learning_rate * grad

print("-" * 62)
print(f"Final w = {w:.4f}, final loss = {loss_fn(w):.6f}")
print(f"Bu w ile noronun ciktisi: {forward(w):.4f}  (hedef: {y_true})")
```

```python
#hedefe ulaıp ulaşmadığını görmek için adım sayısını artırıyorum

learning_rate = 0.5
steps = 300
first_rows = 22   
write = False    

print("=== YONTEM 2: Karpathy usulu (forward difference) ===")
w = -2.0
print(f"{'adim':>4} | {'w':>8} | {'out':>8} | {'loss':>10} | {'grad':>10}")
print("-" * 62)
for step in range(steps):
    z = w * x + b
    out = sigmoid(z)
    loss = loss_fn(w)
    grad = numerical_gradient_karpathy(loss_fn, w)

    if step < first_rows or step == steps - 1:
        print(f"{step:>4} | {w:>8.4f} | {out:>8.4f} | {loss:>10.6f} | {grad:>10.4f}")
    elif not write:
        print("...")
        write = True

    w = w - learning_rate * grad

print("-" * 62)
print(f"Final w = {w:.4f}, final loss = {loss_fn(w):.6f}")
print(f"Bu w ile noronun ciktisi: {forward(w):.4f}  (hedef: {y_true})")
```
