```python
# z = a1 * w1 + a2 * w2 + ... + b   

EULER = 2.718281828459045

def sigmoid(z):
    # Sigmoid fonksiyonu: 1 / (1 + e^-z)
    return 1 / (1 + (EULER ** -z))


def forward_pass(weights, x, bias):
    z = bias
    for i in range(len(weights)):
        z += weights[i] * x[i]
    return z, sigmoid(z)

#ornek
z, out = forward_pass([0.5, -0.2, 0.1], [1.0, 0.5, 0.2], 0.5)
print(f"raw (z) = {z}")
print(f"sigmoid(z) = {out}")

```

```python
# Birden fazla norondan olusan kucuk bir katman kur, forward pass'i buna genislet.
def forward_pass(x, weights, biases):

    outputs = []

    for neuron_weights, bias in zip(weights, biases):
        neuron_output = 0

        for input_value, weight in zip(x, neuron_weights):  
            neuron_output += input_value * weight
        neuron_output += bias
        
        outputs.append(sigmoid(neuron_output))
    return outputs

weights = [[0.5, -0.2], [-0.3, 0.8], [0.7, 0.1]]  # 3 neuron, her biri kendi 2 tane olan agırlıgıyla
biases = [0.1, -0.2, 0.3]                          # 3 neuron, her biri kendi biasıyla
outputs = forward_pass([2.0, 3.0], weights, biases)
print(outputs)
```
